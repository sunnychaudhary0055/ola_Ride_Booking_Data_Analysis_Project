```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel("Bookings-100000-Rows.xlsx")
```

```python
df
```

```python
df.drop_duplicates()
```

```python
df.info()
```

```python
df["V_TAT"] = df["V_TAT"].fillna(0)
```

```python
df["V_TAT"]= pd.to_numeric(df["V_TAT"], errors="coerce")
```

```python
df["C_TAT"] = df["C_TAT"].fillna(0)
```

```python
df.info()
```

```python
df["Canceled_Rides_by_Customer"] = df["Canceled_Rides_by_Customer"].fillna("NaN")
```

```python
df["Canceled_Rides_by_Driver"] = df["Canceled_Rides_by_Driver"].fillna("NaN")
```

```python
df.info()
```

```python
# Ab hum baki columns me jo null values hain unhe ek sath fill karenge

df = df.fillna({
    "Incomplete_Rides" : "NaN",
    "Incomplete_Rides_Reason" : "NaN",
    "Payment_Method" : "NaN",
    "Driver_Ratings" : 0,
    "Customer_Rating" : 0
})
```

```python
df.info()
```

```python
# Dirty Values - N/A, NULL, Blank, spaces

df.replace(                                                 # Data me ["N/A", "NULL", "", " "] ye values jaha bhi hongi unhe "NAN" matlab "Blank" se replace kar dega
    ["N/A", "NULL", "", " "], np.nan, inplace=True          # Ye kiya taki kahi koi null ya blank space bach na jaye
)
```

```python
df.isnull().sum()                   # Kahi koi null value nahi he
```

```python
df = df.apply(lambda x : x.str.strip().str.title() if x.dtype == "String" else x)       # Taki kahi extra ya unwanted space ho to hatjaye or cells ke andar j bhi string likha he wo "Title" wale format me ho jaye
```

```python
df["Pickup_Location"].value_counts()
```

```python
df.describe()
```

```python
df.info()
```

```python
df
```

```python
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S", errors="coerce")
```

```python
df["Time"].dtype
```

```python
df
```

```python
df.info()
```

```python
# Humne dekha ki Booking_Value har jagah dali hui he jabki booking status "Success" ho ya na ho

df["Booking_Value"] = df["Booking_Value"].where(df["Booking_Status"]=="Success", 0)
```

```python
df
```

```python
df.to_csv("Cleaned_Bookings-100000-Rows.xlsx", index=False)
```
